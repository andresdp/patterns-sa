```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
aggregation_df = pd.read_csv('gateway_aggregation_robustness_matrix.csv', index_col=0)
offloading_df = pd.read_csv('gateway_offloading_robustness_matrix.csv', index_col=0)
```

```python
# aggregation_df
```

```python
# offloading_df
```

```python
gateway_df = pd.concat([aggregation_df, offloading_df], axis=1)
gateway_df.fillna(0, inplace=True)
#gateway_df.reset_index(inplace=True)
gateway_df.index.name = 'responsetime_utilization'
gateway_df.columns = ['aggregation', 'offloading']
gateway_df
```

```python
#plt.figure(figsize=(8,8))
#ax = sns.heatmap(df, cmap='BuPu', vmin=0, vmax=1, linewidths=0.5)
sns.set(font_scale=0.8)
ax = sns.clustermap(gateway_df, cmap='BuPu', robust=True, method="average", vmin=0.0, vmax=0.5, figsize=(5,5))
ax.ax_row_dendrogram.set_visible(False)
ax.ax_col_dendrogram.set_visible(False)
#ax.ax_heatmap.set_xticklabels([])

#ax.set_xlabel('configurations')
#ax.set_ylabel('tradeoffs')
#ax.set_xticks([])
#ax.set_yticks([])
plt.suptitle("Robustness of Gateway configurations w.r.t. tradeffs")
plt.show()
```
